# EMPLOYEE PERFORMANCE DATASET

* Age: Age of the employee
* Gender: Gender of the employee
* EducationBackground: Educational qualification
* MaritalStatus: Marital status of the employee
* EmpDepartment: Department in which employee works
* EmpJobRole: Job role assigned to employee
* BusinessTravelFrequency: Frequency of business travel
* DistanceFromHome: Distance between home and workplace
* EmpEducationLevel: Education level (numeric)
* EmpEnvironmentSatisfaction: Satisfaction with work environment
* EmpHourlyRate: Hourly wage of the employee
* EmpJobInvolvement: Level of job involvement
* EmpJobLevel: Job level in organization
* EmpJobSatisfaction: Job satisfaction score
* NumCompaniesWorked: Number of companies previously worked
* OverTime: Whether employee works overtime
* EmpLastSalaryHikePercent: Percentage of last salary hike
* EmpRelationshipSatisfaction: Relationship satisfaction score
* TotalWorkExperienceInYears: Total work experience
* TrainingTimesLastYear: Trainings attended last year
* EmpWorkLifeBalance: Work-life balance rating
* ExperienceYearsAtThisCompany: Years at current company
* ExperienceYearsInCurrentRole: Years in current role
* YearsSinceLastPromotion: Years since last promotion
* YearsWithCurrManager: Years with current manager
Attrition: Whether employee left the company
PerformanceRating: Target variable (employee performance score)
